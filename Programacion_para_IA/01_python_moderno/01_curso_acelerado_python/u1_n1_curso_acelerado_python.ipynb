### Mutabilidad, alias y copia

Dos nombres pueden apuntar a la misma lista. Para modificar el contenedor por separado, debemos copiarlo.
Si contiene otros objetos mutables, una copia superficial aún puede compartirlos. Este detalle explica muchos errores silenciosos
en pipelines de datos.

```python
original = ["dato-a", "dato-b"]
alias = original
copia = original.copy()

alias.append("dato-c")
copia.append("dato-independiente")

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("¿Mismo objeto?", original is alias, original is copia)
```

### Copiar una estructura anidada

`.copy()` hace una **copia superficial**: crea el contenedor externo, pero conserva
las referencias a sus elementos. Si dentro hay una lista, esa lista sigue compartida.

```python
registro_original = {"texto": "hola", "etiquetas": ["saludo"]}
registro_copia = registro_original.copy()
registro_copia["etiquetas"].append("corto")
print(registro_original)
print(registro_original is registro_copia)
print(registro_original["etiquetas"] is registro_copia["etiquetas"])
```

### Ejercicio 3 · Copiar sin compartir la lista

Crea una copia que permita añadir `corto` sin cambiar las etiquetas del original. Para esta estructura basta copiar también la lista interna. **Profundización:** investiga qué cambiaría si hubiera más niveles.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Predicción: copiar el diccionario y su lista de etiquetas separará los
# contenedores mutables; agregar "corto" a la copia no cambiará la fuente.
fuente = {"texto": "hola", "etiquetas": ["saludo"]}
copia_independiente = fuente.copy()
copia_independiente["etiquetas"] = fuente["etiquetas"].copy()
copia_independiente["etiquetas"].append("corto")

print("Fuente:", fuente)
print("Copia:", copia_independiente)
assert fuente == {"texto": "hola", "etiquetas": ["saludo"]}
assert copia_independiente == {"texto": "hola", "etiquetas": ["saludo", "corto"]}
assert fuente is not copia_independiente
assert fuente["etiquetas"] is not copia_independiente["etiquetas"]
# Explicación: .copy() solo copia un nivel. El texto puede compartirse porque
# las cadenas son inmutables, pero la lista necesita su propio contenedor.

# Profundización: con más niveles, una segunda copia superficial podría
# conservar listas o diccionarios internos compartidos. copy.deepcopy recorre
# esos niveles y también copia los contenedores mutables anidados.
from copy import deepcopy

fuente_anidada = {"datos": {"etiquetas": [["saludo"]]}}
copia_profunda = deepcopy(fuente_anidada)
copia_profunda["datos"]["etiquetas"][0].append("corto")
assert fuente_anidada["datos"]["etiquetas"] == [["saludo"]]
assert copia_profunda["datos"]["etiquetas"] == [["saludo", "corto"]]
assert fuente_anidada["datos"] is not copia_profunda["datos"]
assert fuente_anidada["datos"]["etiquetas"][0] is not copia_profunda["datos"]["etiquetas"][0]
```

### Diccionarios: representar registros

Un diccionario asocia claves a valores. `in` comprueba las claves; `.items()`

### Conjuntos: pertenencia, coincidencias y diferencias

Un conjunto conserva valores únicos. Usa `set()` para el conjunto vacío: `{}` crea

```python
permitidas = {"positivo", "neutral", "negativo"}
observadas = {"positivo", "positivo", "desconocido"}
print("Reconocidas:", observadas & permitidas)
print("No reconocidas:", observadas - permitidas)
```

## 3. Decisiones y truthiness

En una condición, Python interpreta como falsos `False`, `None`, cero y las

```python
respuesta_modelo = ""

if not respuesta_modelo:
    estado = "sin respuesta"
elif len(respuesta_modelo) < 20:
    estado = "respuesta breve"
else:
    estado = "respuesta disponible"

print(estado)
```

### Bloques, ciclos y decisiones

Los dos puntos abren un bloque y la sangría indica qué instrucciones pertenecen a
él. Usa cuatro espacios por nivel. `for` repite el bloque para cada elemento;
`range(3)` produce los valores 0, 1 y 2. `continue` salta al siguiente elemento y
`break` termina el ciclo. `while` repite mientras una condición sea verdadera.

```python
for valor in [0.0, None, 0.85]:
    if valor is None:
        continue
    print("Score presente:", valor)

intentos = 0
while intentos < 3:
    intentos += 1
    print("Intento", intentos)
```

### Ejercicio 5 · Cero no es ausencia

Recorre `[None, 0.0, 0.4, 0.95, 0.99]`. Ignora `None` y detente en el primer score mayor o igual a 0.9. Deben mostrarse 0.0, 0.4 y 0.95. ¿Qué se perdería usando `if not score`?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Predicción: se ignorará None, se mostrarán 0.0, 0.4 y 0.95, y el ciclo
# terminará antes de visitar 0.99. Hay que mostrar el score antes del break.
valores_busqueda = [None, 0.0, 0.4, 0.95, 0.99]
scores_mostrados = []

for score in valores_busqueda:
    if score is None:
        continue
    print(score)
    scores_mostrados.append(score)
    if score >= 0.9:
        break

assert scores_mostrados == [0.0, 0.4, 0.95]
assert 0.99 not in scores_mostrados
assert bool(0.0) is False
# Explicación: «if not score» también descartaría 0.0, perdiendo una puntuación
# presente. La comprobación «is None» omite únicamente la ausencia de datos.
```

## 4. Ciclos y comprehensions

Un `for` recorre los elementos de un iterable. En una lista puedes recorrer los
valores directamente; no hace falta construir sus índices. Usa `enumerate` cuando
también necesites la posición y `zip` para avanzar por varias secuencias a la vez.
Por defecto, `zip` se detiene al agotarse la secuencia más corta. Si necesitas
comprobar que tengan la misma longitud, usa `zip(..., strict=True)`.

```python
textos = ["muy bueno", "regular", "excelente"]
scores = [0.82, 0.51, 0.96]

for posicion, (texto, score) in enumerate(zip(textos, scores), start=1):
    print(f"{posicion}. {texto:<12} → {score:.0%}")
```

Las comprehensions sirven para transformaciones pequeñas y legibles. Si la
lógica requiere varias decisiones o efectos secundarios, un ciclo explícito
suele ser mejor.

```python
scores_altos = [score for score in scores if score >= 0.80]
longitudes = {texto: len(texto) for texto in textos}
iniciales = {texto[0] for texto in textos}

print(scores_altos)
print(longitudes)
print(iniciales)
```

### Ejercicio 6 · Transformar y filtrar

Convierte a Fahrenheit las temperaturas mayores o iguales a 25. Usa la fórmula `fahrenheit = celsius * 9 / 5 + 32`. Primero usa un ciclo y después una comprehension. El resultado debe ser `[77.0, 84.2, 87.8]`.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Predicción: solo pasan el filtro 25, 29 y 31 °C; sus conversiones serán
# [77.0, 84.2, 87.8] °F en ambas versiones.
temperaturas = [18, 21, 25, 29, 31]
fahrenheit_ciclo = []

for celsius in temperaturas:
    if celsius >= 25:
        fahrenheit = celsius * 9 / 5 + 32
        fahrenheit_ciclo.append(fahrenheit)

fahrenheit_comprehension = [
    celsius * 9 / 5 + 32 for celsius in temperaturas if celsius >= 25
]
print("Ciclo:", fahrenheit_ciclo)
print("Comprehension:", fahrenheit_comprehension)
assert fahrenheit_ciclo == [77.0, 84.2, 87.8]
assert fahrenheit_comprehension == fahrenheit_ciclo
assert temperaturas == [18, 21, 25, 29, 31]
# Explicación: primero filtramos y luego transformamos cada valor admitido.
# La comprehension expresa estos mismos pasos de forma breve y conserva el orden.
```

## 5. Funciones: nombrar una transformación

Una función bien enfocada reduce repetición y permite probar una idea de forma
aislada. Sus parámetros son entradas; `return` produce una salida. Una función
sin `return` explícito devuelve `None`.

```python
def normalizar_texto(texto, *, minusculas=True):
    '''Elimina espacios externos y opcionalmente convierte a minúsculas.'''
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado


print(normalizar_texto("  Hola, IA  "))
print(normalizar_texto("  Hola, IA  ", minusculas=False))
```

El `*` vuelve explícitos los argumentos posteriores: debemos escribir
`minusculas=False`. En la semana 2 añadiremos anotaciones de tipo y Pydantic;
hoy nos concentramos en el comportamiento.

### Verificar una función con `assert`

`assert condición` comprueba una expectativa. Si es falsa, produce un
`AssertionError`. Lo usaremos para comprobar ejercicios; las entradas externas
se validarán con condiciones y excepciones explícitas.

```python
assert normalizar_texto("  HOLA  ") == "hola"
assert normalizar_texto("  HOLA  ", minusculas=False) == "HOLA"
assert normalizar_texto("") == ""
```

### Ejercicio 7 · Promedio sin inventar datos

Escribe `promedio_o_none(valores)`: devuelve el promedio o `None` si la lista está vacía. Comprueba `[0.0]`, `[]` y `[0.5, 1.0]`. ¿Por qué cero y ausencia necesitan resultados distintos?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Predicción: [0.0] dará 0.0, [] dará None y [0.5, 1.0] dará 0.75.
def promedio_o_none(valores):
    """Devuelve el promedio de una lista, o None si no contiene datos."""
    if not valores:
        return None
    return sum(valores) / len(valores)


assert promedio_o_none([0.0]) == 0.0
assert promedio_o_none([]) is None
assert promedio_o_none([0.5, 1.0]) == 0.75
for muestra in ([0.0], [], [0.5, 1.0]):
    print(f"{muestra!r} -> {promedio_o_none(muestra)!r}")
# Explicación: cero es el promedio de una medición real; None expresa que no
# hay observaciones para calcularlo. No inventamos un cero ni dividimos entre
# cero para la lista vacía. [0.0] es una lista no vacía y sí entra al cálculo.
```

### Ordenar usando una función

`sorted` crea una lista nueva; `.sort()` modifica la lista existente. El argumento
`key` recibe una función que calcula el criterio de orden. Pasamos la función
sin llamarla: Python la aplicará a cada elemento.

```python
evaluaciones = [
    {"modelo": "a", "score": 0.7},
    {"modelo": "b", "score": 0.95},
    {"modelo": "c", "score": 0.8},
]

def obtener_score(evaluacion):
    return evaluacion["score"]

ranking = sorted(evaluaciones, key=obtener_score, reverse=True)
print(ranking)
# lambda permite expresar una función pequeña en el mismo lugar.
assert ranking == sorted(evaluaciones, key=lambda e: e["score"], reverse=True)
```

### De contar a agrupar

Un contador responde «¿cuántos?». Una agrupación conserva los elementos asociados
a cada clave. `defaultdict(list)` crea una lista al acceder a una clave ausente.

```python
from collections import defaultdict

conteo_manual = {}
for etiqueta in etiquetas:
    conteo_manual[etiqueta] = conteo_manual.get(etiqueta, 0) + 1
assert conteo_manual == dict(Counter(etiquetas))

por_etiqueta = defaultdict(list)
for registro in [
    {"etiqueta": "positivo", "score": 0.9},
    {"etiqueta": "neutral", "score": 0.6},
    {"etiqueta": "positivo", "score": 0.8},
]:
    por_etiqueta[registro["etiqueta"]].append(registro["score"])
print(dict(por_etiqueta))
```

Usando `por_etiqueta`, construye un diccionario con el promedio de cada grupo. Comprueba que positivo tiene dos scores y neutral uno. **Profundización:** ordena las etiquetas por promedio descendente.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Predicción: positivo tiene dos scores y promedio cercano a 0.85;
# neutral tiene uno y promedio 0.6. Positivo quedará primero al ordenar.
from math import isclose

promedios_por_etiqueta = {}
for etiqueta, scores_grupo in por_etiqueta.items():
    promedios_por_etiqueta[etiqueta] = promedio_o_none(scores_grupo)

print("Promedios:", promedios_por_etiqueta)
assert len(por_etiqueta["positivo"]) == 2
assert len(por_etiqueta["neutral"]) == 1
assert set(promedios_por_etiqueta) == {"positivo", "neutral"}
assert isclose(promedios_por_etiqueta["positivo"], 0.85)
assert isclose(promedios_por_etiqueta["neutral"], 0.6)
# Explicación: items() proporciona la etiqueta y los scores asociados para
# calcular un promedio por grupo. isclose tolera el pequeño redondeo binario
# de los float; 0.8500000000000001 representa aquí el promedio esperado 0.85.

# Profundización: el criterio de orden es el promedio de cada etiqueta.
etiquetas_ordenadas = sorted(
    promedios_por_etiqueta,
    key=lambda etiqueta: promedios_por_etiqueta[etiqueta],
    reverse=True,
)
print("Etiquetas por promedio descendente:", etiquetas_ordenadas)
assert etiquetas_ordenadas == ["positivo", "neutral"]
```

## 8. POO: objetos que conservan estado

Una **clase** define cómo construir objetos y qué operaciones ofrecen. Cada
**instancia** es un objeto concreto. Sus **atributos** guardan datos y sus
**métodos** son funciones asociadas al objeto.

`__init__` inicializa una instancia; `self` representa esa instancia dentro de los
métodos. Al llamar `contador.agregar()`, Python pasa `contador` como `self`.

```python
class Contador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.total = 0

    def agregar(self):
        self.total += 1

    def reiniciar(self):
        self.total = 0

contador_a = Contador("lote A")
contador_b = Contador("lote B")
contador_a.agregar()
contador_a.agregar()
print(contador_a.nombre, contador_a.total)
print(contador_b.nombre, contador_b.total)
assert contador_b.total == 0
```

## 9. Caso integrador: un registro de predicciones

La clase siguiente conserva predicciones ya normalizadas y combina estructuras de
datos con comportamiento y estado. Cada registro recibe su posición al agregarse.
`resumen` devuelve cantidad, conteos y promedio; el promedio es `None` cuando no
hay registros.


```python
predicciones = [
    {"texto": "Excelente", "etiqueta": "positivo", "score": 0.98},
    {"texto": "Regular", "etiqueta": "neutral", "score": 0.66},
    {"texto": "Llegó roto", "etiqueta": "negativo", "score": 0.93},
    {"texto": "Cero válido", "etiqueta": "neutral", "score": 0.0},
]
```

```python
class RegistroPredicciones:
    def __init__(self, nombre):
        self.nombre = nombre
        self._registros = []

    def agregar(self, registro, posicion):
        guardado = registro.copy()
        guardado["posicion"] = posicion
        self._registros.append(guardado)

    def resumen(self):
        scores = [registro["score"] for registro in self._registros]
        return {
            "cantidad": len(self._registros),
            "conteos": dict(Counter(r["etiqueta"] for r in self._registros)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

lote = RegistroPredicciones("Demostración")
for posicion, registro in enumerate(predicciones):
    lote.agregar(registro, posicion)
print(lote.resumen())
```

### Leer el resultado antes de extenderlo

Deben quedar **4 registros**, conteos de positivo: 1, neutral: 2 y negativo: 1,
y promedio **0.6425**. Este promedio describe los scores; no mide exactitud del
modelo porque no tenemos etiquetas de referencia.

```python
# Predicción: el lote conserva cuatro registros, incluso el score 0.0;
# sus posiciones serán 0, 1, 2 y 3. El promedio vacío será None, no cero.
resumen_lote = lote.resumen()
assert resumen_lote["cantidad"] == 4
assert resumen_lote["conteos"] == {"positivo": 1, "neutral": 2, "negativo": 1}
assert isclose(resumen_lote["promedio"], 0.6425)
assert [registro["posicion"] for registro in lote._registros] == [0, 1, 2, 3]
assert len(lote._registros) == len(predicciones)
assert all("posicion" not in registro for registro in predicciones)
assert all(
    guardado is not original
    for guardado, original in zip(lote._registros, predicciones, strict=True)
)

registros_cero = [registro for registro in lote._registros if registro["score"] == 0.0]
assert registros_cero == [
    {"texto": "Cero válido", "etiqueta": "neutral", "score": 0.0, "posicion": 3}
]

lote_vacio = RegistroPredicciones("Vacío")
resumen_vacio = lote_vacio.resumen()
assert resumen_vacio["cantidad"] == 0
assert resumen_vacio["conteos"] == {}
assert resumen_vacio["promedio"] is None
assert lote_vacio._registros == []
assert lote_vacio._registros is not lote._registros

# Modificación: un lote con una única observación cero sí tiene promedio 0.0.
lote_cero = RegistroPredicciones("Solo cero")
lote_cero.agregar(predicciones[3], 0)
assert lote_cero.resumen() == {
    "cantidad": 1, "conteos": {"neutral": 1}, "promedio": 0.0
}
assert lote.resumen() == resumen_lote
assert lote_vacio.resumen() == resumen_vacio
print("Lote:", resumen_lote)
print("Vacío:", resumen_vacio)
print("Solo cero:", lote_cero.resumen())
# Explicación: la suma 2.57 dividida entre cuatro incluye el cero y da 0.6425.
# Es un promedio de scores; sin etiquetas de referencia no mide exactitud.
# Las posiciones vienen de enumerate desde cero. agregar copia cada diccionario,
# por eso añadir la posición no modifica las predicciones de entrada.

# Cierre: elegiría una lista para predicciones en orden de llegada, una tupla
# para una coordenada (latitud, longitud), un diccionario para los campos de un
# registro y un conjunto para sus etiquetas únicas. Distinguiría cero de
# ausencia con «is None». Una copia superficial aún comparte elementos mutables
# internos; hay que copiarlos también o usar deepcopy para más niveles.
# RegistroPredicciones tiene atributos nombre y _registros; __init__ inicializa,
# agregar almacena y resumen calcula cantidad, conteos y promedio. Dos lotes
# necesitan listas distintas para que las nuevas predicciones de uno no cambien
# al otro. Conservar estado ayuda cuando los registros llegan en varias llamadas.
```
